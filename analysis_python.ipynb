```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_excel("Churn_Unclean_Project.xlsx")
```

```python
#checking missing values
df.isnull().sum()
```

```python
#checking duplicates
df.duplicated().sum()
```

```python
#removing duplicates
df.drop_duplicates(inplace=True)
```

```python

df.duplicated().sum()
df.shape
```

```python
#replace dirty values
df.replace(['N/A','NULL','',' '],np.nan,inplace=True)
```

```python
df.isnull().sum()
```

```python
#removing extra spaces
df = df.apply(lambda x:x.str.strip() if x.dtype == "str" else x)
```

```python
#proper case 
df["Customer_Name"] = df["Customer_Name"].str.title()
df["State"] = df["State"].str.title()
df["City"] = df["City"].str.title()
df["Subscription_Type"] = df["Subscription_Type"].str.title()
df["Contract_Type"] = df["Contract_Type"].str.title()
df["Churn"] = df["Churn"].str.title()
df["Gender"] = df["Gender"].str.title()
```

```python
df["Subscription_Type"].value_counts()
```

```python
df["Gender"].value_counts()
```

```python
#converting numeric column
colm = ["Age","Tenure_Months","Monthly_Charges","Total_Charges"]
for c in colm :
    df[c] = pd.to_numeric(df[c],errors='coerce')
```

```python
df.info()
```

```python
#removing invalid age(putting filter)
df = df[(df['Age']>=18)&(df["Age"]<=100)]
```

```python
df=df[df['Monthly_Charges']>=0]
df=df[df['Total_Charges']>=0]
```

```python
#convert date
df['Last_Interaction_Date'] = pd.to_datetime(df["Last_Interaction_Date"],errors="coerce")
```

```python
#fill missing values
df["Tech_Support"].fillna("No")
df["Payment_Method"].fillna("Unknown")
df["Monthly_Charges"].fillna(df["Monthly_Charges"].mean())
df["Tenure_Months"].fillna(df["Tenure_Months"].mean())
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
#feature enginerring

#Customer lifetime value
df['Customer_Value'] = df["Monthly_Charges"]*df["Tenure_Months"]
```

```python
#monthly revenue
df["Monthly_Revenue"] = df["Monthly_Charges"]
```

```python
# tenure group "0-12","13-24","25-48","49-72"
bins = [0,12,24,48,72]
label =["Bronze","Silver","Gold","Diamond"]
df["Tenure_Group"] = pd.cut(df["Tenure_Months"],bins = bins ,labels = label)
```

```python
df.head()
```

```python
#senior flag
df["Senior_Flag"] = np.where(df["Age"]>=60,"Senior","Adult")
```

```python
#churn flag
df["Churn_Flag"] = df['Churn'].map({'Yes':1,"No":0})
```

```python
#Final check
df.info()
```

```python
df["Last_Interaction_Date"] = df["Last_Interaction_Date"].fillna(df["Last_Interaction_Date"].median())
```
